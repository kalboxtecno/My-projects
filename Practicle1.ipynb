```python
Q1=df["chol"].quantile(0.25)
Q3=df["chol"].quantile(0.75)
IQR=Q3-Q1
lower=Q1-1.5*IQR
upper=Q3+1.5*IQR

outliers=df[
    (df["chol"]<lower)|
    (df["chol"]>upper)
]
print(outliers["chol"])
```

```python
numeric_columns=[
    "age","trestbps","chol","thalach",
    "oldpeak"
]
for column in numeric_columns:
    Q1=df[column].quantile(0.25)
    Q3=df[column].quantile(0.75)

    IQR=Q3-Q1
    lower=Q1-(1.5*IQR)
    upper=Q3-(1.5*IQR)

    outlier=df[
        (df[column]<lower) |
        (df[column]>upper)
    ]
    print(column,":",len(outliers),"outliers")
```

```python
# Inconsistent Values
data=[
    "sex","cp","fbs","restecg","exang","slope",
    "ca","thal","target",

]
for column in data:
    print(column,df[column].unique())
```

```python
# Perform Validation
invalid_age=df[
    (df["age"]<18)|
    (df["age"]>100)
]
print("invalid_age:",len(invalid_age))
```

```python
invalid_sex=df[
    ~df["sex"].isin([0,1])
]
print("invalid sex:",len(invalid_sex))
```

```python
invalid_fbs=df[
    ~df["fbs"].isin([0,1])

]
print("Invalid fbs:",len(invalid_fbs))
```

```python
invalid_target=df[
    ~df["target"].isin([0,1])
]
print("Inavlid Target:",len(invalid_target))
```

```python
validation_result={
    "Invalid age":((df["age"]<18)|(df["age"]>100)).sum(),
    "Invalid sex":(~df["sex"].isin([0,1])).sum(),
    "Invalid cp":(~df["cp"].isin([0,1,2,3])).sum(),
    "Invalid trestbps":(df["trestbps"]<=0).sum(),
    "Invalid chol":(df["chol"]<=0).sum(),
    "Invalid fbs":(~df["fbs"].isin([0,1])).sum(),
    "Invalid restecg":(~df["restecg"].isin([0,1,2])).sum(),
    "Invalid thalach":(df["thalach"]<=0).sum(),
    "Invalid exang":(~df["exang"].isin([0,1])).sum(),
    "Invalid oldpeak":(df["oldpeak"]<0).sum(),
    "Invalid slope":(~df["slope"].isin([0,1,2])).sum(),
    "Invalid ca":(~df["ca"].isin([0,1,2,3])).sum(),
    "Invalid thal":(~df["thal"].isin([0,1,2,3])).sum(),
    "Invalid target":(~df["target"].isin([0,1])).sum()
}
for rule,count in validation_result.items():
    print(rule,":",count)
```
